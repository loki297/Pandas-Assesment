Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing. 

### Given Dataset: `student_assessment_dirty.csv`

The dataset contains the following records:

| Student_ID | Student_Name | Department | Assessment | Marks |
|------------|--------------|------------|------------|-------|
| S101 | Asha | IT | Python | 85 |
| S102 | Ravi | it | Python | 78 |

| S104 | Karan | HR | Python | |

### 1.Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().


```python
import pandas as pd
df = pd.read_csv("student_assessment_dirty.csv")
print("First five rows:")
print(df.head())
print("\nDataset information:")
df.info()
print("\nMissing values:")
print(df.isnull().sum())
```

### Explanation:
First, we load the CSV file using pd.read_csv(). Then, head() is used to view the first few rows, info() is used to check the structure and data types, and isnull().sum() is used to identify the number of missing values in each column. This inspection helps us understand the problems in the dataset before cleaning it

### 2.Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.


```python
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .replace({
        "I.T.": "IT",
        "IT": "IT",
        "FINANCE": "FINANCE",
        "HR": "HR"
    })
)
print(df["Department"].value_counts())
```

### Explanation:
We standardize the department names by converting all names to uppercase and removing unnecessary dots. This makes IT, it, and I.T. consistent as IT, which helps avoid errors during data analysis.

### 3.Identify and remove exact duplicate student assessment records.

```python
before = len(df)

df = df.drop_duplicates()

removed = before - len(df)

print("Records removed:", removed)
```

### Explanation:
We standardize the department names by converting all names to uppercase and removing unnecessary dots. This makes IT, it, and I.T. consistent as IT, which helps avoid errors during data analysis.

#### 4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.


```python
df["Marks"] = (
    df["Marks"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df["Marks"])
```

### Explanation:
The Marks column is converted into a numeric format by extracting the numbers from text values such as "88 marks". Values such as "Absent", "NA", and blank entries cannot be converted into marks, so they are treated as missing values (NaN).

### 5.Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.


```python
print(df[df["Marks"].isna()])
```

### Explanation:

The isna() function is used to identify missing values in the Marks column. The code filters and displays only the student records where marks are not available. This helps us investigate missing marks before deciding whether to remove them or replace them.

### 6.Identify marks outside the valid range of 0–100 and decide how they should be handled.


```python
invalid = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print(invalid)
```

### Explanation:

Identify marks below 0 or above 100, verify the original data, correct data-entry errors, and use NaN if the correct value cannot be determined.

### 7.Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.



```python
print(df)

print("Records removed:", removed)
print("Missing marks remaining:", df["Marks"].isna().sum())
print("Invalid marks identified:", len(invalid))
```

### Explanation:

The cleaned DataFrame shows the data after cleaning. Records removed tells how many rows were deleted. Missing marks remaining tells how many marks are still empty. Invalid marks identified tells how many marks were below 0 or above 100. This helps us check whether the data is clean and correct.
